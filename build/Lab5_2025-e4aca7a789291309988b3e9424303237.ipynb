:::::{admonition} Exercise 2: Computing the Distance to the Cluster M92 
:class: tip

The **horizontal branch** of the HR diagram (see Q4 of Exercise 1) is comprised of stars at roughly the same luminosity regardless of temperature. Since we have a sense of the instrinsic (absolute) luminosity of these stars, we can compared that to their observed luminosity to get a sense of their distance. For a globular cluster like M92, we can then average over the distance to these special stars to get a distance to the cluster. In this exercise, we'll do exactly that!
