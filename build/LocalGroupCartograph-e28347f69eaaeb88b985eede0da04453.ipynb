# Mapping the Local Group of Galaxies

### Context

In this practice, we are going to make a map of the **Local Group** of galaxies (and associated star clusters).  Roughly speaking, the Local Group refers to the gravitationally-bound group of galaxies within which the Milky Way resides; it spans out to roughly 10 million light years from the Milky Way (3 Megaparsecs, in more common units in the field). This volume contains just two massive galaxies -- the Milky Way and Andromeda -- in addition to hundreds of low mass dwarf galaxies and thousands of star clusters as well.


We usually think of the positions of things in the Local Group in **Supergalactic** Cartesian coordinates (X, Y, Z); however, we cannot observe these directly. Instead, we typically record the angular positions of stars/galaxies etc in "Right Ascension" (0-360 degrees) and "Declination" (-90 to 90 degrees) as well as their distances with respect to the Sun (their "heliocentric distances"); these measurable quantities can be converted so long as one assumes a coordinate frame for the Local Group.

For today's exercise, we're going to give you this conversion in the form of a function below.

```python
from astropy import units as u
from astropy.coordinates import SkyCoord
import numpy as np

def convert_to_supergalactic(ra, dec, distance_kpc):
    """
    Convert RA, Dec, and heliocentric distance to Supergalactic Coordinates.
    
    Parameters:
        ra (float): Right Ascension in degrees
        dec (float): Declination in degrees
        distance (float): Heliocentric distance in kpc
        
    Returns:
        tuple: (SGX, SGY, SGZ) in Mpc
    """
    # Create a SkyCoord object
    coord = SkyCoord(ra=ra * u.deg, dec=dec * u.deg, distance=distance * u.kpc, frame='icrs')
    
    # Convert to Supergalactic coordinates
    sg_coord = coord.transform_to('supergalactic')
    
    # Convert to Cartesian coordinates
    sgx = sg_coord.cartesian.x.to(u.Mpc).value
    sgy = sg_coord.cartesian.y.to(u.Mpc).value
    sgz = sg_coord.cartesian.z.to(u.Mpc).value
    
    return sgx, sgy, sgz
```

An example of using this function for a single trio of (RA,DEC,distance):

```python
# Example usage:
ra = 180.0  # degrees
dec = 45.0  # degrees
distance = 10.0  # kpc
sgx, sgy, sgz = convert_to_supergalactic(ra, dec, distance)
print(f"SGX: {sgx:.2f} kpc")
print(f"SGY: {sgy:.2f} kpc")
print(f"SGZ: {sgz:.2f} kpc")
```
